```python
import random
from collections import deque
from IPython.display import HTML, display

# We will render the interactive game using HTML5 Canvas and JavaScript.
# This brings the full GUI experience directly into Colab's output without tkinter!
html_code = """
<div style="font-family: Arial, sans-serif; text-align: center; background: #222; color: #fff; padding: 20px; border-radius: 10px; max-width: 500px; margin: auto;">
    <h2>🤖 AI Arena</h2>
    <div id="stats" style="font-size: 16px; margin-bottom: 10px; font-weight: bold;">
        ❤️ Health: 3 | Cost: 0 | BFS Nodes: 0
    </div>
    <canvas id="gameCanvas" width="450" height="450" style="border: 2px solid #555; background: #111; border-radius: 5px;"></canvas>
    <div style="margin-top: 15px;">
        <button id="startBtn" style="padding: 10px 20px; font-size: 16px; font-weight: bold; background: #2196f3; color: white; border: none; border-radius: 5px; cursor: pointer; margin-right: 15px;">▶ START AI</button>
        <button id="newMapBtn" style="padding: 10px 20px; font-size: 16px; font-weight: bold; background: #4caf50; color: white; border: none; border-radius: 5px; cursor: pointer;">🔄 NEW MAP</button>
    </div>
</div>

<script>
(function() {
    const canvas = document.getElementById('gameCanvas');
    const ctx = canvas.getContext('2d');
    const statsEl = document.getElementById('stats');
    const startBtn = document.getElementById('startBtn');
    const newMapBtn = document.getElementById('newMapBtn');

    const N = 10;
    const C = 45;
    let S = [0, 0];
    let T = [0, 0];
    let W = new Set();
    let H = 3;
    let COST = 0;
    let NODES = 0;
    let path = [];
    let isRunning = false;

    function bfs(start, target, obstacles) {
        let q = [[start]];
        let visited = new Set();
        visited.add(start.toString());

        while (q.length > 0) {
            let currPath = q.shift();
            let node = currPath[currPath.length - 1];
            if (node[0] === target[0] && node[1] === target[1]) {
                return currPath;
            }
            const directions = [[1,0], [-1,0], [0,1], [0,-1]];
            for (let [dx, dy] of directions) {
                let nextNode = [node[0] + dx, node[1] + dy];
                let key = nextNode.toString();
                if (nextNode[0] >= 0 && nextNode[0] < N && nextNode[1] >= 0 && nextNode[1] < N) {
                    if (!obstacles.has(key) && !visited.has(key)) {
                        visited.add(key);
                        q.push([...currPath, nextNode]);
                    }
                }
            }
        }
        return null;
    }

    function generateMap() {
        S = [Math.floor(Math.random() * N), Math.floor(Math.random() * N)];
        H = 3;
        COST = 0;
        NODES = 0;
        while (true) {
            T = [Math.floor(Math.random() * N), Math.floor(Math.random() * N)];
            W = new Set();
            for (let i = 0; i < 25; i++) {
                let r = Math.floor(Math.random() * N);
                let c = Math.floor(Math.random() * N);
                if ((r !== S[0] || c !== S[1]) && (r !== T[0] || c !== T[1])) {
                    W.add([r, c].toString());
                }
            }
            path = bfs(S, T, W);
            if (path) break;
        }
        draw();
    }

    function draw(highlightPath = [], currentPos = null) {
        ctx.clearRect(0, 0, canvas.width, canvas.height);
        let hPathSet = new Set(highlightPath.map(p => p.toString()));

        for (let r = 0; r < N; r++) {
            for (let c = 0; c < N; c++) {
                let key = [r, c].toString();
                if (W.has(key)) {
                    ctx.fillStyle = '#333';
                } else if (hPathSet.has(key)) {
                    ctx.fillStyle = '#90caf9';
                } else {
                    ctx.fillStyle = '#1e1e1e';
                }
                ctx.fillRect(c * C, r * C, C, C);
                ctx.strokeStyle = '#2d2d2d';
                ctx.strokeRect(c * C, r * C, C, C);
            }
        }

        // Draw Target 🎯
        ctx.font = '24px Arial';
        ctx.textAlign = 'center';
        ctx.textBaseline = 'middle';
        ctx.fillText('🎯', T[1] * C + C/2, T[0] * C + C/2);

        // Draw AI Agent
        let activePos = currentPos || S;
        ctx.beginPath();
        ctx.arc(activePos[1] * C + C/2, activePos[0] * C + C/2, C/2 - 6, 0, Math.PI * 2);
        ctx.fillStyle = '#2196f3';
        ctx.fill();
        ctx.closePath();

        statsEl.innerText = `❤️ Health: ${H} | Cost: ${COST} | BFS Nodes: ${NODES}`;
    }

    function runStep() {
        let p = bfs(S, T, W);
        if (!p) {
            statsEl.innerText = `🏆 GAME OVER | Cost: ${COST}`;
            isRunning = false;
            return;
        }
        NODES += p.length;
        COST += p.length - 1;

        let i = 0;
        function animateStep() {
            if (i < p.length) {
                draw(p, p[i]);
                i++;
                setTimeout(animateStep, 120);
            } else {
                S = p[p.length - 1];
                H--;
                if (H > 0) {
                    while (true) {
                        T = [Math.floor(Math.random() * N), Math.floor(Math.random() * N)];
                        if (!W.has(T.toString()) && bfs(S, T, W)) break;
                    }
                    setTimeout(runStep, 200);
                } else {
                    statsEl.innerText = `🏆 GAME OVER | Total Cost: ${COST}`;
                    isRunning = false;
                }
            }
        }
        animateStep();
    }

    startBtn.onclick = () => {
        if (isRunning || H <= 0) return;
        isRunning = true;
        runStep();
    };

    newMapBtn.onclick = () => {
        isRunning = false;
        generateMap();
    };

    generateMap();
})();
</script>
"""
display(HTML(html_code))
```
